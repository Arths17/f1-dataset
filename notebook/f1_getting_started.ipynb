# F1 World Championship (1950-2026): Getting Started

A quick tour of the dataset: loading the core tables, joining results with driver/constructor names, and a look at the current 2026 championship standings.

```python
import os
import pandas as pd

DATA_DIR = None
for root, dirs, files in os.walk("/kaggle/input"):
    if "races.csv" in files:
        DATA_DIR = root
        break
print("Using data dir:", DATA_DIR)

races = pd.read_csv(f"{DATA_DIR}/races.csv")
results = pd.read_csv(f"{DATA_DIR}/results.csv")
drivers = pd.read_csv(f"{DATA_DIR}/drivers.csv")
constructors = pd.read_csv(f"{DATA_DIR}/constructors.csv")
driver_standings = pd.read_csv(f"{DATA_DIR}/driver_standings.csv")

print(f"Races: {len(races)}   Results: {len(results)}   Drivers: {len(drivers)}   Constructors: {len(constructors)}")
```

## Most recent race weekend

```python
races["date"] = pd.to_datetime(races["date"])
latest_race = races.sort_values("date").iloc[-1]
print(f"Most recent race: {latest_race['name']} ({latest_race['date'].date()}) - round {latest_race['round']}, {latest_race['year']}")

top10 = (
    results[results.raceId == latest_race.raceId]
    .merge(drivers[["driverId", "forename", "surname"]], on="driverId")
    .sort_values("positionOrder")
    .head(10)
)
top10[["positionOrder", "forename", "surname", "points", "grid"]]
```

## Current driver championship standings (2026)

```python
races_2026 = races[races.year == 2026]
latest_2026_race_id = races_2026.sort_values("date").iloc[-1].raceId

standings = (
    driver_standings[driver_standings.raceId == latest_2026_race_id]
    .merge(drivers[["driverId", "forename", "surname"]], on="driverId")
    .sort_values("position")
)
standings[["position", "forename", "surname", "points", "wins"]].head(10)
```

## Races per season over time

```python
import matplotlib.pyplot as plt

races_per_year = races.groupby("year").size()
races_per_year.plot(kind="line", figsize=(10, 4), title="Number of F1 races per season, 1950-2026")
plt.ylabel("Races")
plt.xlabel("Season")
plt.show()
```
